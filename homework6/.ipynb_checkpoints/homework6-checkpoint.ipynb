```python
import numpy as np
```

```python
# 3.1

list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

print(arr * 4)

print(np.mean(arr))

print(np.median(arr))

print(np.std(arr))

min_index = np.argmin(arr)
max_index = np.argmax(arr)

print(arr[min_index])
print(arr[max_index])

print(np.min(arr))
print(np.max(arr))
# Yes, they match the values from the previous step

print(arr[arr > 20])
```

```python
# 3.2

matrix = [
    [5, 12, 17, 9], 
    [8, 21, 14, 3], 
    [6, 13, 19, 11]
]
matrix = np.array(matrix)

print(np.shape(matrix))
# there are 3 rows and 4 columns

third_row = matrix[2]
print(third_row)

second_column = matrix[:, 1]
print(second_column)

print(matrix * 2)

print(np.mean(matrix, axis = 0))

print(np.mean(matrix))

print(np.max(matrix, axis = 1))
print(np.min(matrix, axis = 0))

flat_matrix = matrix.flatten()
print(flat_matrix)
print(np.shape(flat_matrix))
# The matrix is now 1D with 12 items

print(np.argmin(flat_matrix))
print(np.argmax(flat_matrix))

print(np.unravel_index(np.argmin(flat_matrix), matrix.shape))
print(np.unravel_index(np.argmax(flat_matrix), matrix.shape))

print(matrix[matrix > 15])

sorted_array = np.sort(matrix, axis = 1)
print(sorted_array)
```

```python
# 4

arr = np.array([
    [2, 3, 5], 
    [14, 6, 8], 
    [11, 13, 17], 
    [7, 10, 13]
])

def is_prime(num):
    if num < 2:
        return False
    for divisor in range(2, int(num / 2)):
        if num % divisor == 0:
            return False
    return True
        
def return_prime_rows(arr):
    prime_rows = []
    for row in arr:
        for num in row:
            if is_prime(num) == True:
                prime_rows.append(row)
                break
    return prime_rows

print(return_prime_rows(arr))
```

```python
# 5.1

def zeros_8x8():
    arr = np.zeros((8, 8))
    return arr
```

```python
# 5.2

def odd_rows_alternating():
    arr = np.zeros((8, 8))
    arr[0::2, ::2] = 1
    return arr
```

```python
# 5.3

def checkerboard():
    arr = np.zeros((8, 8))
    arr[0::2, ::2] = 1
    arr[1::2, 1::2] = 1
    return arr

checkerboard()
```

```python
# 5.4

def reverse_checkerboard():
    arr = np.zeros((8, 8))
    arr[1::2, ::2] = 1
    arr[0::2, 1::2] = 1
    return arr

reverse_checkerboard()
```

```python
# 6 

def expansion(string, num):
    spaces = " " * num
    string = spaces.join(string)
    return string

expansion("universe", 2)
```

```python
# 7

def second_brightest(stars):
    second_brightest_stars = []
    for cluster in stars:
        brightest_index = np.argmax(cluster)
        cluster[brightest_index] = 0
        second_brightest = np.max(cluster)
        second_brightest_stars.append(second_brightest)
    return second_brightest_stars

stars = np.array([[1123, 1456, 1789, 1324, 1876],
[1567, 1987, 1678, 1405, 1589],
[1345, 1654, 1523, 1109, 1923],
[1298, 1890, 1367, 1784, 1432],
[1823, 1756, 1489, 1672, 1550]])
second_brightest(stars)
```
