```python
# The columns represent the different components of each entry, including the state abbreviation, latitude, longitude, and name
# The first state in the data set is AK
# The last state in the dataset is WY
# There are 52 total entries
# There are more than 50 states listed because the data also includes DC and territories
```

```python
med_latitude = df["latitude"].median()
med_longitude = df["longitude"].median()

print(med_latitude)
print(med_longitude)
```

```python
df["med long difference"] = abs(df["longitude"] - med_longitude)
```

```python
print(df.shape)
# There are now 52 rows and 5 columns
```

```python
df["med lat difference"] = abs(df["latitude"] - med_latitude)

closest_lat = df["med lat difference"].min(axis = 0)
farthest_lat = df["med lat difference"].max(axis = 0)
closest_long = df["med long difference"].min(axis = 0)
farthest_long = df["med long difference"].max(axis = 0)

closest_lat_i = df[df["med lat difference"] == closest_lat].index[0]
farthest_lat_i = df[df["med lat difference"] == farthest_lat].index[0]
closest_long_i = df[df["med long difference"] == closest_long].index[0]
farthest_long_i = df[df["med long difference"] == closest_long].index[0]

closest_lat_state = df.loc[closest_lat_i, "state"]
farthest_lat_state = df.loc[farthest_lat_i, "state"]
closest_long_state = df.loc[closest_long_i, "state"]
farthest_long_state = df.loc[farthest_long_i, "state"]

print(f"The state closest to the median latitude is {closest_lat_state}, with a difference of {closest_lat:.2f}.")
print(f"The state farthest to the median latitude is {farthest_lat_state}, with a difference of {farthest_lat:.2f}.")
print(f"The state closest to the median longitude is {closest_long_state}, with a difference of {closest_long:.2f}.")
print(f"The state farthest to the median longitude is {farthest_long_state}, with a difference of {farthest_long:.2f}.")
```

```python
sorted_df = df.sort_values(by = ["med lat difference"], axis = 0, ascending = True)

# I found this function at https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html. 
# The function sorts the rows or columns in a dataframe based on the parameters you input. In this case, I sorted the rows by ascending difference from median latitude. 

sorted_df.head()
```
