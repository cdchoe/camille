```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows = 1)
```

```python
print(data.shape)
# There are 2642 rows and 2 columns
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print(f"The minimum magnitude is {np.min(magnitude)}, which corresponds to star {np.argmin(magnitude)}.")

print(f"The maximum magnitude is {np.max(magnitude)}, which corresponds to star {np.argmax(magnitude)}.")
```

```python
print(f"The minimum temperature is {np.min(temperature)}, which corresponds to star {np.argmin(temperature)}.")

print(f"The maximum temperature is {np.max(temperature)}, which corresponds to star {np.argmax(temperature)}.")
```

```python
# The brightest star is star 1988, with a magnitude of 2.4
# The dimmest star is star 264, with a magnitude of 12.49
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize = (8, 6))

im = plt.scatter(log_temperature, log_magnitude, c = log_temperature, cmap = "RdYlBu")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[1988], log_magnitude[1988], color = "black", marker = "*", s = 250, label = "brightest star")
plt.scatter(log_temperature[264], log_magnitude[264], color = "purple", marker = "*", s = 250, label = "dimmest star")
plt.scatter(log_temperature[2613], log_magnitude[2613], color = "green", marker = "*", s = 250, label = "hottest star")
plt.scatter(log_temperature[1923], log_magnitude[1923], color = "orange", marker = "*", s = 250, label = "coolest star")

cbar = plt.colorbar(im, label = "Temperature (log K)", orientation = "horizontal", pad = 0)
cbar.ax.invert_xaxis()

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.ylabel("Magnitude")
plt.xticks([])

plt.show()
```
