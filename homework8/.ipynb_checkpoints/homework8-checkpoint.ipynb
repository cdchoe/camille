```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sin_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
combined = list(zip(amplitude, frequency))

linestyles = ["-", "--", "-.", ":"]

combined_A = list(zip(amplitude, linestyles))
combined_w = list(zip(frequency, linestyles))
combined = list(zip(combined, linestyles))
```

```python
fig, axes = plt.subplots(1, 3, figsize = (25, 5))

for num, linestyle in combined_A:
    axes[0].plot(x, make_sin_wave(x, num, 1), label = f"A = {num}", ls = linestyle)
    axes[0].set_title("Sine Wave - Varying Amplitude")
    axes[0].set_xlabel("x")
    axes[0].set_ylabel("sin(x)")
    axes[0].legend()

for num, linestyle in combined_w:
    axes[1].plot(x, make_sin_wave(x, 1, num), label = f"w = {num}", ls = linestyle)
    axes[1].set_title("Sine Wave - Varying Frequency")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("sin(x)")
    axes[1].legend()

for ((A, w), linestyle) in combined:
    axes[2].plot(x, make_sin_wave(x, A, w), label = f"(A, w) = {A, w}", ls = linestyle)
    axes[2].set_title("Sine Wave - Varying Amplitude and Frequency")
    axes[2].set_xlabel("x")
    axes[2].set_ylabel("sin(x)")
    axes[2].legend()

plt.show()
```
